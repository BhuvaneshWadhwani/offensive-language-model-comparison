```python
# CLASS WEIGHTS

train_labels = np.array(train["label"])

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(train_labels),
    y=train_labels
)

class_weights = torch.tensor(class_weights, dtype=torch.float)

print("Class weights:", class_weights)
```

```python
# CUSTOM TRAINER WITH CLASS-WEIGHTED LOSS
class WeightedLossTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels = inputs.pop("labels")

        outputs = model(**inputs)
        logits = outputs.logits

        loss_fct = torch.nn.CrossEntropyLoss(
            weight=class_weights.to(logits.device)
        )

        loss = loss_fct(logits, labels)

        return (loss, outputs) if return_outputs else loss
```

```python
# CUSTOM DECODER TRAINER WITH CLASS-WEIGHTED LOSS
class WeightedDecoderTrainer(Trainer):
    """
    Causal-LM Trainer that reweights the loss by class - the decoder
    equivalent of WeightedLossTrainer. Since the decoder has no single
    classification logit, the "class" lives in the answer tokens
    (e.g. "Offensive"/"Non-Offensive") inside an otherwise -100-masked
    `labels` sequence. We compute per-token loss unreduced, average it
    per example, then scale each example's loss by its class weight
    before averaging across the batch.

    Requires each example to carry its raw integer class label under
    `class_label` (added to decoder_train_ds below).
    """

    def __init__(self, *args, class_weights=None, label_column="class_label", **kwargs):
        super().__init__(*args, **kwargs)
        self.class_weights = class_weights.to(self.args.device) if class_weights is not None else None
        self.label_column = label_column

    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        class_labels = inputs.pop(self.label_column)

        outputs = model(**inputs)
        logits = outputs.logits          # [batch, seq_len, vocab_size]
        lm_labels = inputs["labels"]     # [batch, seq_len], -100 at masked positions

        # Standard causal-LM shift: token i's logits predict token i+1
        shift_logits = logits[:, :-1, :].contiguous()
        shift_labels = lm_labels[:, 1:].contiguous()

        loss_fct = torch.nn.CrossEntropyLoss(reduction="none", ignore_index=-100)
        token_losses = loss_fct(
            shift_logits.view(-1, shift_logits.size(-1)),
            shift_labels.view(-1)
        ).view(shift_labels.size())

        # Average per-token loss within each example (so answer length doesn't distort weighting), then apply the class weight
        mask = (shift_labels != -100).float()
        per_example_loss = (token_losses * mask).sum(dim=1) / mask.sum(dim=1).clamp(min=1)

        if self.class_weights is not None:
            example_weights = self.class_weights[class_labels]
            loss = (per_example_loss * example_weights).sum() / example_weights.sum()
        else:
            loss = per_example_loss.mean()

        return (loss, outputs) if return_outputs else loss
```

```python
# METRICS

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)

    accuracy = accuracy_score(labels, predictions)

    precision, recall, weighted_f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="weighted"
    )

    macro_f1 = f1_score(labels, predictions, average="macro")

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "weighted_f1": weighted_f1,
        "macro_f1": macro_f1
    }


def count_parameters(model):
    trainable_params = sum(
        p.numel() for p in model.parameters() if p.requires_grad
    )

    total_params = sum(
        p.numel() for p in model.parameters()
    )

    trainable_pct = trainable_params / total_params * 100

    return trainable_params, total_params, trainable_pct


def print_encoder_evaluation(model_name, true_labels, pred_labels, label_names, model):
    accuracy = accuracy_score(true_labels, pred_labels)
    macro_f1 = f1_score(true_labels, pred_labels, average="macro")
    per_class_f1 = f1_score(true_labels, pred_labels, average=None)

    trainable_params, total_params, trainable_pct = count_parameters(model)

    summary_df = pd.DataFrame({
        "Metric": [
            "Accuracy",
            "Macro F1",
            f"F1 {label_names[0]}",
            f"F1 {label_names[1]}",
            "Trainable parameters",
            "Total parameters",
            "Trainable %"
        ],
        "Value": [
            accuracy,
            macro_f1,
            per_class_f1[0],
            per_class_f1[1],
            trainable_params,
            total_params,
            trainable_pct
        ]
    })

    print(f"\n{model_name}")
    print("=" * 70)

    display(summary_df)

    print("\nClassification report:")
    print(
        classification_report(
            true_labels,
            pred_labels,
            target_names=label_names
        )
    )

    cm = confusion_matrix(true_labels, pred_labels)

    print("Confusion matrix:")
    print(cm)

    plt.figure(figsize=(5, 4))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        xticklabels=label_names,
        yticklabels=label_names
    )

    plt.xlabel("Predicted label")
    plt.ylabel("True label")
    plt.title(f"{model_name} Confusion Matrix")
    plt.tight_layout()
    plt.show()
```

### 2.2 Decoder Model

```python
# QWEN 2 MODEL SETUP

# Model Configuration
decoder_model_name = "Qwen/Qwen3-0.6B"
device = "cuda" if torch.cuda.is_available() else "cpu"

print(f"Loading decoder model: {decoder_model_name} on device: {device}")

decoder_tokenizer = AutoTokenizer.from_pretrained(decoder_model_name)

if decoder_tokenizer.pad_token is None:
    decoder_tokenizer.pad_token = decoder_tokenizer.eos_token

decoder_model = AutoModelForCausalLM.from_pretrained(
    decoder_model_name,
    torch_dtype="auto",
    device_map="auto" if torch.cuda.is_available() else None
)

print("Decoder model and tokenizer loaded successfully.")
print(f"Model parameters: {decoder_model.num_parameters():,}")
print("Labels:", label_names)
```

```python
# FUNCTIONS

def count_parameters(model):
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    total_params = sum(p.numel() for p in model.parameters())
    trainable_pct = trainable_params / total_params * 100

    return trainable_params, total_params, trainable_pct


def generate_decoder_prediction(text, decoder_model, decoder_tokenizer):
    messages = [
        {
            "role": "user",
            "content": (
                "Classify the following tweet as exactly one label: "
                "Offensive or Non-Offensive.\n\n"
                f"Tweet: {text}\n\n"
                "Answer with only the label."
            )
        }
    ]

    formatted_text = decoder_tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
        enable_thinking=False
    )

    model_inputs = decoder_tokenizer(
        [formatted_text],
        return_tensors="pt",
        max_length=128,
        truncation=True
    ).to(decoder_model.device)

    with torch.inference_mode():
        generated_ids = decoder_model.generate(
            **model_inputs,
            max_new_tokens=10,
            do_sample=False,
            pad_token_id=decoder_tokenizer.eos_token_id,
            eos_token_id=decoder_tokenizer.eos_token_id,
        )

    input_length = model_inputs.input_ids.shape[1]
    output_ids = generated_ids[0][input_length:]

    prediction = decoder_tokenizer.decode(
        output_ids,
        skip_special_tokens=True
    ).strip()

    return prediction


def decoder_prediction_to_label(prediction):
    pred = prediction.lower().strip()

    if "non-offensive" in pred or "non offensive" in pred:
        return 0
    elif "offensive" in pred:
        return 1
    else:
        return 0


def print_decoder_evaluation(model_name, true_labels, pred_labels, label_names, model):
    accuracy = accuracy_score(true_labels, pred_labels)
    macro_f1 = f1_score(true_labels, pred_labels, average="macro")
    per_class_f1 = f1_score(true_labels, pred_labels, average=None)

    trainable_params, total_params, trainable_pct = count_parameters(model)

    summary_df = pd.DataFrame({
        "Metric": [
            "Accuracy",
            "Macro F1",
            f"F1 {label_names[0]}",
            f"F1 {label_names[1]}",
            "Trainable parameters",
            "Total parameters",
            "Trainable %"
        ],
        "Value": [
            accuracy,
            macro_f1,
            per_class_f1[0],
            per_class_f1[1],
            trainable_params,
            total_params,
            trainable_pct
        ]
    })

    print(f"\n{model_name}")
    print("=" * 70)
    display(summary_df)

    print("\nClassification report:")
    print(classification_report(true_labels, pred_labels, target_names=label_names))

    cm = confusion_matrix(true_labels, pred_labels)

    print("Confusion matrix:")
    print(cm)

    plt.figure(figsize=(5, 4))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        xticklabels=label_names,
        yticklabels=label_names
    )

    plt.xlabel("Predicted label")
    plt.ylabel("True label")
    plt.title(f"{model_name} Confusion Matrix")
    plt.tight_layout()
    plt.show()


def evaluate_decoder_model(decoder_model, model_display_name, eval_dataset, decoder_tokenizer, num_samples=None):
    if num_samples is None:
        num_samples = len(eval_dataset)

    predictions = []
    pred_labels = []
    true_labels = []

    print(f"\nEvaluating {model_display_name} on {num_samples} samples...")

    for i in range(min(num_samples, len(eval_dataset))):
        sample = eval_dataset[i]

        prediction = generate_decoder_prediction(
            sample["text"],
            decoder_model,
            decoder_tokenizer
        )

        pred_label = decoder_prediction_to_label(prediction)
        true_label = sample["label"]

        predictions.append(prediction)
        pred_labels.append(pred_label)
        true_labels.append(true_label)

        if (i + 1) % 10 == 0:
            print(f"Processed {i + 1}/{num_samples} samples...")

        if i < 3:
            print(f"\nExample {i + 1}:")
            print(f"Text: {sample['text']}")
            print(f"True: {label_names[true_label]}")
            print(f"Predicted text: {prediction}")
            print(f"Predicted label: {label_names[pred_label]}")

    print_decoder_evaluation(
        model_display_name,
        true_labels,
        pred_labels,
        label_names,
        decoder_model
    )

    return predictions, pred_labels, true_labels
```

```python
# INSTRUCTION FINE-TUNING DATASET PREPARATION

def prepare_decoder_instruction_dataset(dataset, decoder_tokenizer, max_length=512):
    # Right-padding is required here so the manual label masking below
    # (which fills labels left-to-right) lines up correctly with input_ids.
    decoder_tokenizer.padding_side = "right"

    all_input_ids = []
    all_labels = []
    label_names = dataset.features["label"].names

    print(f"Preparing {len(dataset)} decoder instruction examples...")

    for idx, row in enumerate(dataset):
        label_text = label_names[row["label"]]

        prompt_messages = [
            {
                "role": "user",
                "content": (
                    "Classify the following tweet as exactly one label: "
                    "Offensive or Non-Offensive.\n\n"
                    f"Tweet: {row['text']}\n\n"
                    "Answer with only the label."
                )
            }
        ]

        full_messages = prompt_messages + [
            {
                "role": "assistant",
                "content": label_text + decoder_tokenizer.eos_token
            }
        ]

        # Prompt-only text (everything the model reads before it has to
        # answer) - only used here to measure how many tokens to mask out;
        # it is never fed to the model on its own.
        prompt_text = decoder_tokenizer.apply_chat_template(
            prompt_messages,
            tokenize=False,
            add_generation_prompt=True,
            enable_thinking=False
        )

        full_text = decoder_tokenizer.apply_chat_template(
            full_messages,
            tokenize=False,
            add_generation_prompt=False,
            enable_thinking=False
        )

        prompt_ids = decoder_tokenizer(prompt_text)["input_ids"]
        full_ids = decoder_tokenizer(
            full_text, truncation=True, max_length=max_length
        )["input_ids"]

        # Mask everything up to the end of the prompt with -100, so only
        # the answer tokens (label word + eos) contribute to the loss -
        # the model is never trained to "predict" the fixed instruction text.
        prompt_len = min(len(prompt_ids), len(full_ids))
        labels = [-100] * prompt_len + full_ids[prompt_len:]

        all_input_ids.append(full_ids)
        all_labels.append(labels)

        if idx == 0:
            print("\nFirst formatted decoder training example:")
            print(full_text)

    # Pad input_ids/attention_mask using the tokenizer's own padding logic
    padded = decoder_tokenizer.pad(
        {"input_ids": all_input_ids},
        padding=True,
        max_length=max_length,
        return_tensors="pt",
    )
    input_ids = padded["input_ids"]
    attention_mask = padded["attention_mask"]

    # Pad labels to the same shape; padding positions also get -100
    padded_labels = torch.full_like(input_ids, -100)
    for i, lab in enumerate(all_labels):
        padded_labels[i, :len(lab)] = torch.tensor(lab, dtype=torch.long)

    final_dataset = Dataset.from_dict({
        "input_ids": input_ids.tolist(),
        "attention_mask": attention_mask.tolist(),
        "labels": padded_labels.tolist(),
    })

    return final_dataset


decoder_train_ds = prepare_decoder_instruction_dataset(
    train,
    decoder_tokenizer,
    max_length=128
)

decoder_eval_ds = prepare_decoder_instruction_dataset(
    validation,
    decoder_tokenizer,
    max_length=128
)

# Keep the true integer label alongside the tokenized instruction text,
# so WeightedDecoderTrainer can look up each example's class weight
decoder_train_ds = decoder_train_ds.add_column("class_label", train["label"])
decoder_eval_ds = decoder_eval_ds.add_column("class_label", validation["label"])

print("Decoder instruction datasets prepared.")
print(f"Training samples: {len(decoder_train_ds)}")
print(f"Evaluation samples: {len(decoder_eval_ds)}")
```

```python
# LoRA FINE-TUNING

# Configuring LoRA

decoder_lora_config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    inference_mode=False,

    r=16,
    lora_alpha=32,
    lora_dropout=0.1,

    target_modules=[
        "q_proj",
        "k_proj",
        "v_proj",
        "o_proj",
        "gate_proj",
        "up_proj",
        "down_proj",
    ],
)

print("Creating decoder LoRA model...")

decoder_lora_model = get_peft_model(decoder_model, decoder_lora_config)
decoder_lora_model.print_trainable_parameters()


# Use mixed precision only when CUDA is available so the notebook also runs on CPU.
decoder_training_args = TrainingArguments(
    output_dir="./decoder_lora_results",

    num_train_epochs=2,
    learning_rate=1e-4,

    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    gradient_accumulation_steps=4,

    warmup_steps=50,

    fp16=torch.cuda.is_available(),

    logging_steps=10,
    report_to=[],

    eval_strategy="steps",
    eval_steps=50,

    save_strategy="steps",
    save_steps=100,

    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    greater_is_better=False,

    remove_unused_columns=False,
    dataloader_pin_memory=False,
    dataloader_drop_last=True,

    max_grad_norm=1.0,
)

decoder_data_collator = default_data_collator

decoder_trainer = WeightedDecoderTrainer(
    model=decoder_lora_model,
    args=decoder_training_args,
    train_dataset=decoder_train_ds,
    eval_dataset=decoder_eval_ds,
    data_collator=decoder_data_collator,
    class_weights=class_weights,
)

print("Starting decoder LoRA instruction fine-tuning...")

decoder_training_output = decoder_trainer.train()

print("Decoder training completed.")
```

```python
# pick example tweets

def select_examples(true_labels, pred_labels, texts, class_id, num_correct=1, num_incorrect=1):
    """ 
    Pick a few example indices for a given class
    """
    true_labels = np.array(true_labels)
    pred_labels = np.array(pred_labels)

    class_mask = true_labels == class_id
    correct_mask = class_mask & (pred_labels == true_labels)
    incorrect_mask = class_mask & (pred_labels != true_labels)

    correct_idx = np.where(correct_mask)[0][:num_correct]
    incorrect_idx = np.where(incorrect_mask)[0][:num_incorrect]

    selected = list(correct_idx) + list(incorrect_idx)
    return [(i, texts[i], true_labels[i], pred_labels[i]) for i in selected]
```

```python
def clean_tokens_for_display(tokens):
    """Replace BPE space/newline markers (RoBERTa's 'Ġ'/'Ċ') with real
    spaces/newlines so tokens render as readable text instead of
    garbled 'Gword' artifacts."""
    return [t.replace("Ġ", " ").replace("Ċ", "\n") for t in tokens]


# INTEGRATED GRADIENTS FOR ENCODER

from captum.attr import LayerIntegratedGradients

def encoder_forward(input_ids, attention_mask):
    """ 
    Captum differentiates this function's output with respect to the input embeddings, 
    so we need to define a forward function that returns the logits for the target class.
    """

    outputs = encoder_model(input_ids=input_ids, attention_mask=attention_mask)
    return outputs.logits

lig_encoder = LayerIntegratedGradients(encoder_forward, encoder_model.roberta.embeddings)

def explain_encoder(text, target_class):
    """ 
    Per-token attibution scores explaining why the encoder leaned toward 'target_class' for this tweet    
    """

    encoding = encoder_tokenizer(
        text, 
        return_tensors='pt', 
        truncation=True,
        max_length=128
    ).to(device)
    input_ids = encoding["input_ids"]
    attention_mask = encoding["attention_mask"]

    # Same sequence for baseline but filled with [PAD]
    # IG integrates gradients along the straight  line from this baseline to the real input.
    baseline_ids = torch.full_like(input_ids, encoder_tokenizer.pad_token_id)
    baseline_ids[0, 0] = input_ids[0, 0]
    baseline_ids[0, -1] = input_ids[0, -1]

    attributions, delta = lig_encoder.attribute(
        inputs=input_ids,
        baselines=baseline_ids,
        additional_forward_args=(attention_mask,),
        target=int(target_class),
        return_convergence_delta=True,
        n_steps=50
    )

    # Collapse the embedding dimension to get a single attribution score per token
    scores = attributions.sum(dim=-1).squeeze(0)  # collapse embedding dim, not token dim
    scores = scores / torch.norm(scores) # normalize for comparable visualization

    tokens = encoder_tokenizer.convert_ids_to_tokens(input_ids.squeeze(0))
    tokens = clean_tokens_for_display(tokens)  # replace BPE space/newline markers for readable display
    return tokens, scores.detach().cpu().numpy(), delta
```

```python
# VISUALIZE ENCODER EXAMPLES

from captum.attr import visualization as viz

def render_attribution(tokens, scores, pred_label, true_label, delta):
    record = viz.VisualizationDataRecord(
        word_attributions=scores,
        pred_prob=1.0,  # not used in this context
        pred_class=label_names[pred_label],
        true_class=label_names[true_label],
        attr_class=label_names[pred_label],
        attr_score=scores.sum(),
        raw_input_ids=tokens,
        convergence_score=delta
    )
    viz.visualize_text([record])

for idx, text, true, pred in encoder_examples:
    tokens, scores, delta = explain_encoder(text, target_class=pred)
    print(f"\nTweet [{idx}]: {text}")
    render_attribution(tokens, scores, pred, true, delta)
```

```python
# INTEGRATED GRADIENTS FOR DECODER
 
label_first_token_id = {}
for lbl in label_names:
    ids = decoder_tokenizer(lbl, add_special_tokens=False)["input_ids"]
    label_first_token_id[lbl] =ids[0]

def build_decoder_prompt(text):
    messages = [{
        "role": "user",
        "content": (
            "Classify the following tweet as exactly one label: "
            "Offensive or Non-Offensive.\n\n"
            f"Tweet: {text}\n\n"
            "Answer with only the label."
        )
    }]
    return decoder_tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True, enable_thinking=False
    )

def decoder_forward(input_ids, attention_mask):
    """
    Captum differentiates this function's output with respect to the input embeddings, 
    so we need to define a forward function that returns the logits for the target class.
    """

    outputs = decoder_lora_model(
        input_ids=input_ids,
        attention_mask=attention_mask
    )
    return outputs.logits[:, -1, :]  # return logits for the last token only

embedding_layer = decoder_lora_model.get_input_embeddings() 
lig_decoder = LayerIntegratedGradients(decoder_forward, embedding_layer)

def explain_decoder(text, target_label_name):
    prompt = build_decoder_prompt(text)
    encoding = decoder_tokenizer(
        prompt, 
        return_tensors="pt", 
        truncation=True, 
        max_length=128
    ).to(decoder_lora_model.device)
    input_ids = encoding["input_ids"]
    attention_mask = encoding["attention_mask"]

    baseline_ids = torch.full_like(input_ids, decoder_tokenizer.pad_token_id)
    target_token_id = label_first_token_id[target_label_name]

    attributions, delta = lig_decoder.attribute(
        inputs=input_ids,
        baselines=baseline_ids,
        additional_forward_args=(attention_mask,),
        target=int(target_token_id),
        return_convergence_delta=True,
        n_steps=50
    )

    scores = attributions.sum(dim=-1).squeeze(0)
    scores = scores / torch.norm(scores)

    tokens = decoder_tokenizer.convert_ids_to_tokens(input_ids.squeeze(0))
    tokens = clean_tokens_for_display(tokens)  # same cleaning function defined in the encoder cell above
    return tokens, scores.detach().cpu().numpy(), delta
```

```python
# TRAINABLE PARAMETERS - ENCODER AND DECODER

from peft import PeftModel

def count_parameters(model):
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    total_params = sum(p.numel() for p in model.parameters())
    trainable_pct = trainable_params / total_params * 100
    return trainable_params, total_params, trainable_pct


# --- Encoder (fully fine-tuned) ---
enc_trainable, enc_total, enc_pct = count_parameters(encoder_model)
print("ENCODER (RoBERTa-base)")
print(f"  Trainable parameters: {enc_trainable:,}")
print(f"  Total parameters:     {enc_total:,}")
print(f"  Trainable %:          {enc_pct:.4f}%")

print()

# --- Decoder (LoRA) ---
decoder_lora_model = PeftModel.from_pretrained(
    decoder_base_model,
    "./lora_offensive_classifier",
    is_trainable=True
)

dec_trainable, dec_total, dec_pct = count_parameters(decoder_lora_model)
print("DECODER (Qwen3-0.6B + LoRA)")
print(f"  Trainable parameters: {dec_trainable:,}")
print(f"  Total parameters:     {dec_total:,}")
print(f"  Trainable %:          {dec_pct:.4f}%")

print()

# --- Summary comparison ---
size_ratio = dec_total / enc_total
param_ratio = enc_trainable / dec_trainable
print(f"Decoder base model is {size_ratio:.1f}x larger than the encoder.")
print(f"Encoder has {param_ratio:.1f}x more trainable parameters than the LoRA decoder.")
```
